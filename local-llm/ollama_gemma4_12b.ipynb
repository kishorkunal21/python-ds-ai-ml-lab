```python
from langchain_ollama import ChatOllama

llm  = ChatOllama(
    model="gemma4:12b",
    base_url="http://localhost:11500"
)

response = llm.invoke("are you connected?")

print(response.content)
```

```python
llm.invoke("how many params you're trained on? can u change code on my system if asked ?")
```

```python
from langchain_ollama import ChatOllama

llm  = ChatOllama(
    model="gemma4:latest",
    base_url="http://127.0.0.1:11500"
)

response = llm.invoke("are you connected?")

print(response.content)
```
